```python
import pysr as PySRegressor
```

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
np.random.seed(123)
n = 200
x = np.random.uniform(-3,3,n)
epsilon = np.random.normal(loc =0, scale = 0.3, size = n)
y= x**2 + 2*np.sin(x) + epsilon

plt.figure(figsize = (8,5))
plt.scatter(x,y, alpha = 0.6)
plt.title("Datos simulados")
plt.show()
```

```python
from pysr import PySRRegressor

X = x.reshape(-1, 1)

model = PySRRegressor(
    niterations=20,
    binary_operators=[
        "+",
        "-",
        "*"
    ],
    unary_operators=[
        "sin"
    ],
    verbosity=1
)

model.fit(X, y)
```

```python
model.equations_
```

```python
model.sympy()
```

```python
X_grid = np.linspace(-3,3,500)
X_grid = X_grid.reshape(-1,1)

y_pred = model.predict(X_grid)
plt.figure(figsize = (8,5))
plt.scatter(x, y, alpha = 0.5)
plt.plot(X_grid,y_pred, linewidth = 2, label = "Regresión simbolica")
plt.legend()
plt.show()
```
