# Maquinas de soporte vectorial

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.datasets import make_moons, make_circles
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import Pipeline
from sklearn.svm import SVC
from sklearn.metrics import accuracy_score
from sklearn.model_selection import (
    train_test_split,
    KFold,
    cross_val_score,
    GridSearchCV,
    RandomizedSearchCV
)

```

```python
def graficar_modelo(
    modelo,
    X,
    y,
    titulo="Frontera de decisión"
):
    
    # Crear figura
    plt.figure(figsize=(8, 6))
    
    # Límites de la región que vamos a visualizar
    x_min = X[:, 0].min() - 0.8
    x_max = X[:, 0].max() + 0.8
    
    y_min = X[:, 1].min() - 0.8
    y_max = X[:, 1].max() + 0.8
    
    # Crear una malla de puntos
    xx, yy = np.meshgrid(
        np.linspace(x_min, x_max, 400),
        np.linspace(y_min, y_max, 400)
    )
    
    # Convertir la malla en pares (X1, X2)
    grid = np.c_[
        xx.ravel(),
        yy.ravel()
    ]
    
    # Predecir la clase de cada punto de la malla
    Z_pred = modelo.predict(grid)
    
    # Regresar al formato de la malla
    Z_pred = Z_pred.reshape(xx.shape)
    
    # Colorear las regiones según la clase predicha
    plt.contourf(
        xx,
        yy,
        Z_pred,
        alpha=0.15
    )

    # Evaluar la función de decisión en toda la malla
    Z_decision = modelo.decision_function(grid)
    
    # Esta función está pensada para clasificación binaria
    Z_decision = Z_decision.reshape(xx.shape)
    
    # Sólo incluimos niveles que realmente estén dentro
    # del rango de la función de decisión
    niveles = [
        nivel
        for nivel in [-1, 0, 1]
        if Z_decision.min() <= nivel <= Z_decision.max()
    ]
    
    # Dibujamos los niveles únicamente si existen
    if len(niveles) > 0:
        
        estilos = [
            "-" if nivel == 0 else "--"
            for nivel in niveles
        ]
        
        plt.contour(
            xx,
            yy,
            Z_decision,
            levels=niveles,
            linestyles=estilos
        )
    
    for clase in np.unique(y):
        
        plt.scatter(
            X[y == clase, 0],
            X[y == clase, 1],
            label=f"Clase {clase}",
            s=60
        )
    

    plt.xlabel("$X_1$")
    plt.ylabel("$X_2$")
    plt.title(titulo)
    plt.legend()
    plt.show()
```

```python
X, y = make_moons(n_samples = 250, noise = 0.2, random_state = 42)
y = np.where(y==0, -1,1)
plt.figure(figsize = (8,6))

for clase in np.unique(y):
    plt.scatter(
        X[y==clase, 0],
        X[y==clase, 1],
        label =f"Clase {clase}"
    )
plt.xlabel("$X_1$")
plt.ylabel("$X_2$")
plt.legend()
plt.show()
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X,y, test_size = 0.3, random_state = 42, stratify = y
)

modelo_lineal = Pipeline([
    ("scaler", StandardScaler()),
    ("svc", SVC(kernel = "linear", C =1))
    ]
)

modelo_lineal.fit(X_train,y_train)


scaler_ajustado = modelo_lineal.named_steps["scaler"]
media = scaler_ajustado.mean_
desviacion = scaler_ajustado.scale_
pred_ajustado = modelo_lineal.named_steps["svc"]
coeficientes =pred_ajustado.coef_[0]
intercepto = pred_ajustado.intercept_[0]

def prediction1(X):
    scalex0 = (np.array(X[:,0])+media[0])/desviacion[0]
    scalex1 = (np.array(X[:,1])+media[1])/desviacion[1]
    pred = intercepto + scalex0*coeficientes[0] + scalex1*coeficientes[1]
    res = []
    for i in range(0,len(pred)):
        if pred[i]<0:
            res.append(-1)
        else:
            res.append(1)
    return res
    
pred2 = prediction1(X_test)   
```

```python
X_train[:3]
scaler_ajustado.transform(X_train[:3])
```

```python
prediccion_pipeline=modelo_lineal.predict(X_test)

print(pred2==prediccion_pipeline)
print(accuracy_score(prediccion_pipeline, y_test))
```

```python
graficar_modelo(modelo_lineal, X_train, y_train)
```

```python
modelo_poly2 = Pipeline([
    (
        "scaler",
        StandardScaler()
    ),
    (
        "svc",
        SVC(
            kernel="poly",
            degree=2,
            C=1,
            gamma="scale",
            coef0=1
        )
    )
])

modelo_poly2.fit(X_train,y_train)
graficar_modelo(modelo_poly2, X_train,y_train, titulo = "kernel polinomial")
```

```python
accuracy_score(y_train,modelo_poly2.predict(X_train))
```

```python
accuracy_score(y_test, modelo_poly2.predict(X_test))
```

```python

for i in range(1,6):
    modelo_poly2 = Pipeline([
        (
            "scaler",
            StandardScaler()
        ),
        (
            "svc",
            SVC(
                kernel="poly",
                degree=i,
                C=1,
                gamma="scale",
                coef0=1
            )
        )
    ])
    
    modelo_poly2.fit(X_train,y_train)
    graficar_modelo(modelo_poly2, X_train,y_train, titulo = f"kernel polinomial de grado {i}")
    print("Accuracy entrenamiento",accuracy_score(y_train,modelo_poly2.predict(X_train)))
    print("Accuracy test",accuracy_score(y_test, modelo_poly2.predict(X_test)))
    
```

```python
modelo_rbf = Pipeline(
    [
        ("scaler",StandardScaler()),
        ("svc",
        SVC(kernel = "rbf",C=1,gamma =1)
        )
        
    ]
)

modelo_rbf.fit(X_train,y_train)
accuracy_score(y_train, modelo_rbf.predict(X_train))
accuracy_score(y_test, modelo_rbf.predict(X_test))
```

```python
graficar_modelo(modelo_rbf,X_train,y_train, titulo = "SVM con kernel radial")
```

```python
valores_gamma = [0.01,0.1,1,10,100]
for gamma in valores_gamma:
    modelo = Pipeline(
        [
            ("scaler", StandardScaler()),
            ("svc", SVC(kernel="rbf",C=1, gamma=gamma))
        ]
    )
    modelo.fit(X_train,y_train)
    print("Accuracy entrenamiento: ",accuracy_score(y_train, modelo.predict(X_train)))
    print("Accuracy test: ",accuracy_score(y_test, modelo.predict(X_test)))
    graficar_modelo(modelo,X_train,y_train, titulo = f"RBF c=1, gamma = {gamma}")
```

```python
valores_gamma = [0.01,0.1,1,10,100]
for gamma in valores_gamma:
    modelo = Pipeline(
        [
            ("scaler", StandardScaler()),
            ("svc", SVC(kernel="rbf",C=gamma, gamma=1))
        ]
    )
    modelo.fit(X_train,y_train)
    print("Accuracy entrenamiento: ",accuracy_score(y_train, modelo.predict(X_train)))
    print("Accuracy test: ",accuracy_score(y_test, modelo.predict(X_test)))
    graficar_modelo(modelo,X_train,y_train, titulo = f"RBF c={gamma}, gamma = 1")
```

```python
cv = KFold(n_splits = 2,
          shuffle= True,
          random_state = 42)
param_grid = {
    "svc__kernel": ["rbf"],
    "svc__C": [0.1, 1, 10],
    "svc__gamma": [0.1, 1, 10]
}

modelo = Pipeline(
    [
        ("scaler", StandardScaler()),
        ("svc", SVC())
    ]
)

grid_search = GridSearchCV(
    estimator=modelo,
    param_grid=param_grid,
    scoring="accuracy",
    cv=cv,
    n_jobs=1,
    return_train_score=True
)

grid_search.fit(X_train, y_train)
```

```python
print(grid_search.best_params_)
print(grid_search.best_score_)
```

```python
X_circulos, y_circulos = make_circles(
    n_samples = 300,
    factor = 0.4,
    noise = 0.10,
    random_state=42
)
y_circulos = np.where(y_circulos ==0,-1,1)
plt.figure()
for clase in np.unique(y_circulos):
    plt.scatter(
        X_circulos[y_circulos == clase, 0],
        X_circulos[y_circulos == clase, 1],
        label = f"Clase = {clase}",
        s=55
               )
plt.legend()
plt.show()
```

```python
Xc_train, Xc_test,yc_train,yc_test = train_test_split(X_circulos,y_circulos, test_size = 0.3,
                                                      random_state=42,stratify = y_circulos)

```

```python
modelo_circulo_lineal = Pipeline(
    [
        ("scale", StandardScaler()),
        ("svc", SVC(kernel = "linear",C=1))
    ]
)
modelo_circulo_lineal.fit(Xc_train,yc_train)
print(accuracy_score(yc_train,modelo_circulo_lineal.predict(Xc_train)))
graficar_modelo(modelo_circulo_lineal,Xc_train,yc_train)
```

```python
modelo_circulo_lineal = Pipeline(
    [
        ("scale", StandardScaler()),
        ("svc", SVC(kernel = "rbf",C=1,gamma =1))
    ]
)
modelo_circulo_lineal.fit(Xc_train,yc_train)
print(accuracy_score(yc_train,modelo_circulo_lineal.predict(Xc_train)))
graficar_modelo(modelo_circulo_lineal,Xc_train,yc_train)
```
